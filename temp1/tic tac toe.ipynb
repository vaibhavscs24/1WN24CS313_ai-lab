```python
import math

# Tic-Tac-Toe Game with AI (Minimax)

HUMAN = 'O'
COMPUTER = 'X'
EMPTY = ' '

def print_board(board):
    for i in range(3):
        print(f" {board[3*i]} | {board[3*i+1]} | {board[3*i+2]} ")
        if i < 2:
            print("---|---|---")
    print()

def is_moves_left(board):
    return EMPTY in board

def evaluate(board):
    # Check rows for victory
    for row in range(3):
        if board[3*row] == board[3*row+1] == board[3*row+2] != EMPTY:
            return 10 if board[3*row] == COMPUTER else -10

    # Check columns for victory
    for col in range(3):
        if board[col] == board[col+3] == board[col+6] != EMPTY:
            return 10 if board[col] == COMPUTER else -10

    # Check diagonals for victory
    if board[0] == board[4] == board[8] != EMPTY:
        return 10 if board[0] == COMPUTER else -10
    if board[2] == board[4] == board[6] != EMPTY:
        return 10 if board[2] == COMPUTER else -10

    return 0

def minimax(board, depth, is_max):
    score = evaluate(board)

    if score == 10:
        return score - depth
    if score == -10:
        return score + depth
    if not is_moves_left(board):
        return 0

    if is_max:
        best = -math.inf
        for i in range(9):
            if board[i] == EMPTY:
                board[i] = COMPUTER
                best = max(best, minimax(board, depth + 1, False))
                board[i] = EMPTY
        return best
    else:
        best = math.inf
        for i in range(9):
            if board[i] == EMPTY:
                board[i] = HUMAN
                best = min(best, minimax(board, depth + 1, True))
                board[i] = EMPTY
        return best

def find_best_move(board):
    best_val = -math.inf
    best_move = -1
    for i in range(9):
        if board[i] == EMPTY:
            board[i] = COMPUTER
            move_val = minimax(board, 0, False)
            board[i] = EMPTY
            if move_val > best_val:
                best_move = i
                best_val = move_val
    return best_move

def play_game():
    board = [EMPTY] * 9
    print("Welcome to Tic-Tac-Toe!")
    print("You are 'O' and the computer AI is 'X'.")
    print("Enter board positions from 1 to 9 (row by row):\n")
    print_board(['1', '2', '3', '4', '5', '6', '7', '8', '9'])

    # Ask who goes first
    turn = input("Do you want to go first? (y/n): ").strip().lower()
    is_human_turn = True if turn == 'y' else False

    while is_moves_left(board) and evaluate(board) == 0:
        if is_human_turn:
            print_board(board)
            move = -1
            while move not in range(9) or board[move] != EMPTY:
                try:
                    user_input = int(input("Enter your move (1-9): ")) - 1
                    if user_input in range(9) and board[user_input] == EMPTY:
                        move = user_input
                    else:
                        print("Invalid move. The slot is either taken or out of range. Try again.")
                except ValueError:
                    print("Please enter a valid number (1-9).")
            board[move] = HUMAN
            is_human_turn = False
        else:
            print("Computer's turn (AI is thinking...)")
            move = find_best_move(board)
            board[move] = COMPUTER
            is_human_turn = True

    print_board(board)
    score = evaluate(board)
    if score > 0:
        print("Computer wins! Game Over.")
    elif score < 0:
        print("Congratulations! You win!")
    else:
        print("It's a draw!")

if __name__ == '__main__':
    play_game()
```
